## Task 1 – Dataset Cleaning
- Load the Titanic dataset (train.csv).
- Display the first 10 rows.
- Check for missing values in each column.
- Fill missing values in the "Age" column with the mean age.
- Drop rows where the "Embarked" column is missing.
## Task 2 – Encoding Categorical Data
- Convert the "Sex" column into numeric (0 = Male, 1 = Female).
- Apply One-Hot Encoding on the "Embarked" column.
## Task 3 – Feature Scaling & Splitting
- Select features: Age, Fare, Sex, Pclass.
- Apply StandardScaler to normalize them.
- Split data into 80% training and 20% testing.


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
df = pd.read_csv("train.csv")
print("--- First 10 Rows ---")
print(df.head(10))
```

```python
print("\n--- Missing Values Before Cleaning ---")
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df = df.dropna(subset=["Embarked"])
```

```python
print("\n--- Missing Values After Cleaning ---")
print(df[["Age", "Embarked"]].isnull().sum())
```

```python
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})
df = pd.get_dummies(df, columns=["Embarked"], prefix="Embarked", dtype=int)

print("\n--- Data Sample After Encoding ---")
print(df[["Sex"] + [col for col in df.columns if "Embarked_" in col]].head())
```

```python
feature_cols = ["Age", "Fare", "Sex", "Pclass"]
X = df[feature_cols]
y = df["Survived"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_train_scaled = pd.DataFrame(
    X_train_scaled, columns=feature_cols, index=X_train.index
)
X_test_scaled = pd.DataFrame(
    X_test_scaled, columns=feature_cols, index=X_test.index
)
print("\n--- Scaled Feature Summary (Training Set) ---")
print(X_train_scaled.describe().round(3))
```
